### Loops

```python
## for loop
for i in range(5):
    print(i)
```

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(1,10,2): ## increment 2 per iteration
    print(i)
```

```python
for i in range(10,0,-1): ## increment 1 per iteration for -1 and give 10 to 1
    print(i)
```

```python
## string

str = "Final Flash"

for i in str:
    print(i)
```

```python
## while loop

count = 1

while count<=5:
    print(count)
    count = count + 1
```

```python
## break statement
for i in range(10):
    if i == 5:
        break ## break the condition by using break statement
    print(i)
```

```python
## continue statement
## continue statement skip the current iteration and continues with the next.

for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
## pass
## pass statement does nothing.(null operation)
for i in range(5):
    if i == 3:
        # print("the number is passed",i)
        pass
    print(i)
```

```python
## nested loop

for i in range(3):
    for j in range(4):
        print(f"i:{i} and j:{j}")
```

```python
## example 1

count = 1
sum = 0

while count <= 10:
    sum = sum+count
    count+= 1
    print(sum)
print(f"total sum:{sum}")
    
```
