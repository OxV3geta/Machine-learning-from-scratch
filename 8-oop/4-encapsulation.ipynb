### Protected Variable

```python
class Person:
    def __init__(self,name,age,gender):
        self._name = name ## protected variable ( '__' makes this private)
        self._age = age   ## protected variable ( '__' makes this private)
        self.gender = gender

class Employee(Person):
        def __init__(self,name,age,gender):
                super().__init__(name,age,gender)


employee = Employee("Vegeta",46,'Male')
print(employee._name)
```

```python
## Encapsulation with Getter and Setter
class Saiyan:
    def __init__(self,name,age):
        self.__name = name ## Private access modifier or variable
        self.__age  = age ## Private access modifier or variable
        
    ## getter method for name
    def get_name(self):
            return self.__name
        
    ## setter method for name
    def set_name(self):
            return self.__name
    
    ## getter method for age 
    def get_age(self):
            return self.__age
    
    ## setter method for age
    def set_age(self,age):
        if age > 0:
            self.__age = age
        else:
            print("Age can't be negative.")
        
person = Saiyan("Vegeta","46")

## Access and modify private variable using getter and setter

print(person.get_name())
print(person.get_age())

person.set_age(40)
print(person.get_age())

person.set_age(-1)

```
