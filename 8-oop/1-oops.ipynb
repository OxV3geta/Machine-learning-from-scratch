```python
### Instance variable and methods

class Dog:
    ##  constructor
    def __init__(self,name,age):
        self.name = name
        self.age = age
        
## create object
dog1 = Dog("niga",3)  
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
dog2 = Dog("lucy",7)
print(dog2.name)
print(dog2.age)
```

```python
## define a class with instance methods

class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):
        print(f"{self.name} says woof and age is {self.age}")
        
dog1=Dog("shiro",4)
dog1.bark()
dog2=Dog("lucy",3)
dog2.bark()
```

```python
class Bank_Account:
    def __init__(self,user,balance):
        self.user = user
        self.balance = balance
    
    def deposite(self,amount):
        self.balance+=amount
        print(f"{amount} added,The new Balance is {self.balance}")
        
    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient balance")
        else:
            self.balance-=amount
            print(f"{amount} is withdrawn,new balance is {self.balance}")
    
    def get_balance(self):
        return self.balance

ac = Bank_Account("Siam",300)
print(ac.user)
print(ac.balance)
        
```
