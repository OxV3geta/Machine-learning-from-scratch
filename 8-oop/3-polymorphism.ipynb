```python
class shape:
    def area(self):
        return f"the area of the figure"

class rectangle(shape):
    def __init__(self,height,width):
        self.height = height
        self.width = width
        
    def area(self):
        return self.height*self.width
    
class circle(shape):
    def __init__(self,radius):
        self.radius = radius
    
    def area(self):
        return 3.1416*self.radius**2
    
def print_area(shape):
    return f"the area is {shape.area()}"

x = rectangle(7,5)
y = circle(3)

print(print_area(x))
print(print_area(y))
```

```python
## Base Class
class Animal:
    def speak(self):
        return "sound of the Animal."

class Dog(Animal):
    def speak(self):
        return "woof!"

class Cat(Animal):
    def speak(self):
        return "Meow!"

def animal_speak(Animal):
    print(Animal.speak())
    
dog = Dog()
cat = Cat()

animal_speak(dog)
print(dog.speak())
print(cat.speak())
```

### polymorphism with abstract base classes
An abstract class is a restricted blueprint that cannot be instantiated directly. It serves as a base for other classes, defining common behaviors while forcing subclasses to provide specific implementations for incomplete (abstract) methods. It acts as the anchor for polymorphism.

```python
from abc import ABC,abstractmethod

class Vehicle:
    @abstractmethod
    def start_engine(self):
        print("the engine started")

class Car(Vehicle):
    def start_engine(self):
        return "car engine started!"
    
class Motorcycle(Vehicle):
    def start_engine(self):
        return "motorcycle engine started!"

def start_vehicle(vehicle):
    print(vehicle.start_engine())
    
car = Car()
start_vehicle(car)
mt = Motorcycle()
start_vehicle(mt)
```
