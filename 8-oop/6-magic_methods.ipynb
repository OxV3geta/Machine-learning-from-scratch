### Magic Method
Magic methods (also known as dunder methods) in Python are special methods that begin and end with double underscores (e.g., __init__, __add__)They allow you to define custom behavior for your classes so that objects interact seamlessly with Python’s built-in operators and functions.


```python
"""
__init__(self, *args, **kwargs) : The actual creation of the object instance (runs before __init__)
__str__(self) - str(obj), print(obj) : Return a string representation of an object.
__eq__(self, other) : Equality check (==) .
__repr__ : Returns an official string representation of an object.

"""
```

```python
class Person:
    pass

person = Person()
dir(person)
```

```python
### Basic magic method
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Name:{self.name} and age:{self.age}"
    
    def __repr__(self):
        return f"Person(name={self.name},age={self.age})"

person = Person("Vegeta",47)
print(person)
print(repr(person))
```
