### Inheritance in python

```python
## Inheritance (single Inheritance)
## Parent class

class Car:
    def __init__(self,type,doors,enginetype):
        self.type = type
        self.doors = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"The person will drive the {self.enginetype} car")
        
```

```python
car1=Car("Suv",5,'petrol')
print(car1.drive())
```

```python
class Tesla(Car):
    def __init__(self,type,doors,enginetype,is_selfdriving):
        super().__init__(type,doors,enginetype)
        self.is_selfdriving = is_selfdriving
    
    def car_info(self):
        print(f"This is {self.type} car with {self.doors} door and the engine type of {self.enginetype} also supported {self.is_selfdriving}")
        
```

```python
tesla_1 = Tesla("sports",5,'electric',True)
tesla_1.car_info()
```

```python
class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        print("Subclass must implement this method.")
    
class Pet():
    def __init__(self,owner):
        self.owner = owner
    
class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speak(self):
        return f"{self.name} say woof!"
    
dog = Dog("shiro","Shinchan")
print(dog.name,dog.owner)
```
