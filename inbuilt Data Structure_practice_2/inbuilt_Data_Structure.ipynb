```python
def sum_num(numbers):
    if not numbers:
        return None
    
    total = 0
    
    for num in numbers:
        total +=num
    return total

print(sum_num([1, 2, 3, 4, 5]))
    
```

```python
def max_num(numbers):
    if not numbers:
        return None
    
    largest = numbers[0]
    for num in numbers[1:]:
        if largest<num:
            largest=num
    
    return largest

print(max_num([3, 8, 2, 10, 5]))
        
```

```python
def remove_duplicates(lst):
    unique_list = []
    
    for item in lst:
        if item not in unique_list:
            unique_list.append(item)
    return unique_list

print(remove_duplicates([1, 2, 2, 3, 4, 4, 5]))
```

```python
def check_list(lst):
    seen = set()
    
    for num in lst:
        if num in seen:
            return False
        seen.add(num)
    
    return True

print(check_list([1, 2, 3, 4, 5]))

            
```

```python
def check_unique(lst):
    seen = set()
    
    for num in lst:
        if num in seen:
            return False
  
        seen.add(num)
    return True
```

```python
def reverse_list(lst):
    start = 0
    end = len(lst) - 1
    
    while start < end:
        lst[start],lst[end] = lst[end],lst[start]
        
        start+=1
        end-=1
        
    return lst
print(reverse_list([1, 2, 3, 4, 5, 6, 7]))
```

```python
def merge_two_sorted_lists(lst1,lst2):
    i=0
    j=0
    result = []
    
    while i < len(lst1) and j < len(lst2):
        if lst1[i] < lst2[j]:
            result.append(lst1[i])
            i+=1
        else:
            result.append(lst2[j])
            j+=1
    
    while i < len(lst1):
        result.append(lst1[i])
        i+=1
    while j < len(lst2):
        result.append(lst2[j])
        j+=1
    
    return result

print(merge_two_sorted_lists([1, 3, 5],[2, 4, 6]))
```

```python
def rotate_list(lst, k):
   
    if not lst:
        return []
    

    k = k % len(lst)
    
    for _ in range(k):
        last_element = lst.pop()  # Remove the last element
        lst.insert(0, last_element)  # Insert it at the front
    
    return lst
print(rotate_list([1, 2, 3, 4, 5], k = 4))
```

```python
def merge_lists_to_dictionary(keys, values):
    # Boundary Condition to check for equal length of keys and values.
    # Last point in problem solving framework :)
    if(len(keys) != len(values)):
        return False
    # Create an empty dictionary to store the result
    result = {}

    # Use a loop to iterate through both lists
    for i in range(len(keys)):
        # Add each key-value pair to the dictionary
        result[keys[i]] = values[i]

    return result
```

```python
def list_to_dictionary(keys,value):
    if(len(keys)!=len(value)):
        return False
    
    result = {}
    
    for i in range(len(keys)):
        result[keys[i]] = value[i]
    return result

print(list_to_dictionary(['a', 'b', 'c'], [1, 2, 3]))
```

```python
def merge_dict(dict1,dict2,dict3):
    merge_dictionary = dict1.copy()
    
    for key,value in dict2.items():
        merge_dictionary[key] = value
        
    for key,value in dict3.items():
        merge_dictionary[key] = value
        
    return merge_dictionary

print(merge_dict({'a': 1, 'b': 2},{'c': 3, 'd': 4},{'e': 5, 'f': 6}))
```

```python
def count_word_frequency(sentence):
    
    word_count = {}
    words = sentence.split()
    
    for word in words:
        if word in word_count:
            word_count[word] +=1
        else:
            word_count[word] = 1
    
    return word_count

print(count_word_frequency("hello world hello"))
    
```

```python
def is_palindromic_tuple(tup):
    start = 0
    end = len(tup) - 1
    
    while start <= end:
        if tup[start] != tup[end]:
            return False
        
        start+=1
        end-+1
    
    return True
print(is_palindromic_tuple((1, 2, 3, 2, 1)))
    
    
```
