```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("This is a message")
logging.error("This is a error")
logging.warning("This is a warning")
logging.info("This is an info message")
logging.critical("This is a critical message")
```

```python
import logging

logging.basicConfig(
    filename='app_logdata.log',
    filemode='w',
    level= logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt= '%y-%m-%d %H:%M:%S'    
)

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message.........")
```

```python
import logging

logger_1 = logging.getLogger("module1")
logger_1.setLevel(logging.DEBUG)

logger_2 = logging.getLogger("module2")
logger_2.setLevel(logging.DEBUG)

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt= '%y-%m-%d %H:%M:%S'
)
```

```python
logger_1.debug("This is debug message for module1")
logger_2.warning("This is a warning message for module2")
logger_2.error("This is an error message")
```
