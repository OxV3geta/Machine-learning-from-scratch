### Python logging
Python logging is a built-in module that allows developers to track events, debug errors, and monitor application behavior by recording messages with specific severity levels during program execution.

```python
import logging

## Configure the basic logging settings
logging.basicConfig(level=logging.DEBUG)

## log message

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
import logging

logging.basicConfig(
    filename = 'app.log',
    filemode='w',
    level = logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s', #s: Convert whatever value is found into a String (text) so it can be printed out.
    datefmt= '%y-%m-%d %H:%M:%S '
)
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```
