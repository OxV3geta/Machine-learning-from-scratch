### Function
Function is a block of code that performs a speific task.it's help in ogranizing code,resuing code and improving readability.

```python
num = 23
if num%2 == 0:
    print("even")
else:
    print("odd")
```

```python
def even_or_odd(num):
    """This funtion finds even or odd"""
    if num%2 == 0:
        print("even")
    else:
        print("odd")
    
```

```python
even_or_odd(26)
```

```python
## funtion with multiple perameters
def add(a,b):
    return a+b

result=add(45,3)

print(result)
```

```python
## Default funtion

def greet(name="prima"):
    print(f"hey cuty {name},Welcome")
    
greet("Maria")
greet()
```

```python
### Variable Length arguments
## Positional and keywords arguments

def print_numbers(*bruh):
    for number in bruh:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"bruh")
```

```python
## positional argument

def print_numbers(*args):
    for numbers in args:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"bruh")
```

```python
## keywords arguments 
def print_details(**kwargs):
    for key,values in kwargs.items():
        print(f"{key}:{values}")
         
```

```python
print_details(name="Vegeta",age=47,breed="Saiyan")
```

```python
## positional and keyword argment combine 
def print_details(*args,**kwargs):
    for numbers in args:
        print(f"positional arguments:{numbers}")
    
    for key,values in kwargs.items():
        print(f"{key}:{values}")
```

```python
print_details(1,2,3,4,5,6,7,8,"bruh",name="Vegeta",age=47,breed="Saiyan")
```
