### Lambda Function
A lambda function is a small, anonymous function defined without a name.They can have any number of arguments but only one expression.

```python
#syntax
lambda arguments: expression
```

```python
def add(x,y):
    return x+y
add(2,5)
```

```python
multiplication = lambda a,b:a*b
type(multiplication)
print(multiplication(5,6))
```

```python
def even(num):
    if num%2==0:
        return True

even(24)
```

```python
even1 = lambda num : num%2==0
even1(13)
```

```python
addition = lambda x,y,z:(x+y+z)
addition(5,5,5)
```

```python
## map() - applies a function to all items in a list
numbers = [1,2,3,4,5,6,7]

def squre(n):
    return n**2

squre(2)
```

```python
## map() - applies a function to all items in a list
numbers = [1,2,3,4,5,6,7]
num = list(map(lambda x:x**2,numbers))
print(num)
```
