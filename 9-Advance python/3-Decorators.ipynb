### Decorator
A Python decorator is a function that takes another function as an argument, extends or modifies its behavior, and returns a new function without permanently modifying the original code.

```python
## function copy
## closures
## decorators
```

```python
## function copy
def welcome():
    return "Welcome to the world!"
```

```python
welcome()
```

```python
wel = welcome
print(wel())

del welcome
print(wel())
```

```python
## closures
def  main_welcome(msg):
    
    def sub_welcome_method():
        print("Brother we are niga!")
        print(msg)
        print("We are very proude of it!")
    return sub_welcome_method()
    
```

```python
main_welcome("hello niger,welcome")
```

```python
def  main_welcome(func,lst):
    
    def sub_welcome_method():
        print("Brother we are niga!")
        print(func(lst))
        print("We are very proude of it!")
    return sub_welcome_method()
    
```

```python
def  main_welcome(func):
    
    def sub_welcome_method():
        print("Brother we are niga!")
        func()
        print("We are very proude of it!")
    return sub_welcome_method()
    
```

```python
def course_introduction():
    print("Messi is the only one Goat.")
    
course_introduction()
```

```python
main_welcome(course_introduction)
```

```python
@main_welcome
def course_introduction():
    print("Messi is the only one Goat.")
```

```python
def decorator(func):
    
    def warpper():
        print("Messi is the Goat!")
        func()
        print("Argentina will win the worldcup.")
    return warpper 
```

```python
@decorator
def say_vamos():
    print("Vamos Argentina")
```

```python
say_vamos()
```

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for i in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
    
```

```python
@repeat(4)
def say_hello():
    print("Goat Msssi")
```

```python
say_hello()
```
