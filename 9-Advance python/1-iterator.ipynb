### Iterators
Iterators in Python are objects that enable sequential traversal of a collection, returning one element at a time without loading the entire dataset into memory.  They are the underlying mechanism powering for loops, comprehensions, and tuple unpacking by implementing the iterator protocol

```python
my_list = [1,2,3,4,5,6]

for i in my_list:
    print(i)
```
