### Generators
Generators in Python are special functions that produce a sequence of values one at a time using the yield keyword, rather than computing and storing all values in memory at once. This mechanism enables lazy evaluation, allowing code to generate items only when requested, which significantly reduces memory usage and improves performance when processing large datasets or infinite sequences.

```python
def squre(n):
    for i in range(7):
        yield (i**2)
        
```

```python
squre(4)
```

```python
for i in squre(2):
    print(i)
```

```python
a = squre(3)
a
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
    
```

```python
gen = my_generator()
gen
```

```python
next(gen)
```

### Practical Example: Reading Large Files

Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
### Practical Example: Reading Large Files

def read_large(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path = 'large.txt'

for line in read_large(file_path):
    print(line.strip())
    
```
