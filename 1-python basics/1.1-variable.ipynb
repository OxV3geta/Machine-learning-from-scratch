### Type Conversion

```python
age = 25
print(type(age))

# type conversion

age_str = str(age)
print(age_str)
print(type(age_str))
```

```python
cg = 3.80
type(cg)
```

```python
float((int(cg)))
```

```python
## Dynamic Typing
## Python allows the type of a variable to change as the program executes

var = 10
print(var,type(var))

var = "khame hame haa"
print(var,type(var))

var = 10.69
print(var, type(var))
```

```python
## input

age = int(input("Enter your CG:"))
print(age,type(age))
```
