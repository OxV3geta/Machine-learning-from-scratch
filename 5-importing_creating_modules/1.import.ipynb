### import Modules in python: Modules and packages

```python
import math

math.sqrt(18)
```

```python
from math import sqrt,pi

print(sqrt(16))
print(sqrt(25))
print(pi)

```

```python
import numpy as np
np.array([1,2,3,4,5])
```

```python
from math import * ## '*' means importing all the function that available in  math or other library/module.

print(sqrt(81))
print(pi)
```

```python
from package.maths import addition

addition(2,3)
```

```python
from package import maths
maths.addition(2,4)
```
