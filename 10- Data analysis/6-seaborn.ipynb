### Data Visualization with seaborn
Seaborn is a Python data visualization library built on top of Matplotlib that integrates closely with Pandas data structures.

```python
import seaborn as sns
```

```python
### Basic plotiing with seaborn
tips = sns.load_dataset('tips')
tips
```

```python
## create a scatter plot

import matplotlib.pyplot as plt

sns.scatterplot(x = 'total_bill', y = 'tip',data=tips)
plt.title('Scatter plot of Total bill vs Tip')
plt.show()
```

```python
## Line Plot

sns.lineplot( x = 'size',y = 'total_bill' , data=tips)
plt.title('Line plot of total bill')
plt.show()
```

```python
## categorical plots
## Bar plot

sns.barplot( x = 'day',y = 'total_bill',data=tips)
plt.title('Bar plot of Total bill by Day')
plt.show()
```

```python
## Box Plot
sns.boxplot(x='day',y='total_bill',data=tips)
```

```python
## Violin Plot

sns.violinplot( x='day',y='total_bill',data=tips)
```

```python
### Histogram
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
# ### KDE plot
# sns.kdeplot(tips['total_bill'],fill=True)

sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## pair plot

sns.pairplot(tips)
```

```python
## Heatmap
tips = sns.load_dataset('tips')
tips
```

```python
## Heatmap
corr = tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd

sales_df = pd.read_csv('Chocolate_Sales.csv').head(10)
sales_df
```

```python
## plot total sales by Product

plt.figure(figsize=(20,10))
sns.barplot(x='Country',y='Amount',data=sales_df,estimator=sum)
plt.title('total sales by Product')
plt.xlabel('Country')
plt.ylabel('Amount')
plt.show()
```

```python
## plot total sales by Product

plt.figure(figsize=(20,10))
sns.barplot(x='Product',y='Amount',data=sales_df,estimator=sum)
plt.title('total sales by Product')
plt.xlabel('Product')
plt.ylabel('Amount')
plt.show()
```
