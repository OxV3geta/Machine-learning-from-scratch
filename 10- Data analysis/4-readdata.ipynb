### Read Data from Different Sources

```python
import pandas as pd
from io import StringIO

Data = '{"Employee name":"James","email":"james@gmail.com","job_profile":[{"title1":"Team Lead","title2":"Sr.Developer"}]}'
df = pd.read_json(StringIO(Data))
```

```python
df.to_json()
```

```python
df.to_json(orient='index')
```

```python
df.to_json(orient='records')
```

```python
url = 'https://en.wikipedia.org/wiki/Mobile_country_code'
headers = {'User-Agent': 'Mozilla/5.0'}

# Read directly using storage_options
tables = pd.read_html(url,match="Country",storage_options=headers)
tables[0]
```
