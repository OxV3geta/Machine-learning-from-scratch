### Pandas
Pandas is an open-source Python library designed for data manipulation and analysis, particularly for handling structured, tabular data similar to spreadsheets or SQL tables. 

In Pandas, the two fundamental data structures are Series and DataFrame, which serve as the building blocks for data manipulation and analysis. 

Series is a one-dimensional, labeled array-like object that can hold any data type (integers, strings, floats, etc.).  It consists of a single column of data with an associated index, functioning similarly to a NumPy array but with explicit labels for each element. Series elements are typically homogeneous, and it supports element-wise computations. 

DataFrame is a two-dimensional, tabular data structure consisting of rows and columns.  It can be viewed as a collection of multiple Series objects or a spreadsheet/SQL table. Unlike Series, DataFrames are heterogeneous, meaning different columns can contain different data types (e.g., numbers, strings, dates).  They support more complex operations, including column-wise computations, alignment, and multi-variable data handling.

```python
## Series
## A pandas Seres is a one-demensional array-like object that can hold any data type.It is similar to a column in a table.

import pandas as pd
data = [3,4,5,1,2]
series = pd.Series(data)
print("Series\n",series)
print(type(series))
```

```python
## create a series of from dictionary
data = {'a':1,'b':4,'c':8}
series_dict = pd.Series(data)

print(series_dict)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
## Dataframe
# create a dataframe from a dictionary of list

data = {
    'name':['Goku','Vegeta','Gohan'],
    'age' :[42,47,22],
    'planet':['Planet Vegeta','Sadala','Earth']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## create a Data frame from list of dictionary

data2 = [
    {'name':'Goku','age' :42,'planet':'Planet Vegeta'},
    {'name':'Vegeta','age' :47,'planet':'salada'},
    {'name':'Gohan','age' :22,'planet':'Earth'},
    {'name':'Piccolo','age' :35,'planet':'namek'}
]
df_2 = pd.DataFrame(data2)
print(df_2)
print(type(df_2))
```
