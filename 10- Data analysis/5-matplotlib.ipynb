### Data Visualization with Matplotlib
Matplotlib is a comprehensive, open-source Python library designed for creating static, animated, and interactive visualizations.  Originally developed by John D. Hunter in 2003, it serves as a foundational plotting tool built over NumPy arrays, offering an object-oriented API for embedding plots into applications.

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]

## creating a line plot

plt.plot(x,y)
plt.xlabel('X label')
plt.ylabel('Y label')
plt.title('Basic Line plot')
plt.show()
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]

### create customized line plot

plt.plot(x,y,color = 'green',linestyle='-.',marker='o',linewidth=3,markersize=8)
plt.grid(True)
```

```python
## Multiple plots
## Sample data

x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title('plot 1')

plt.subplot(2,2,2)
plt.plot(y1,y2,color='blue')
plt.title('plot 2')


plt.subplot(2,2,3)
plt.plot(x,y2,color='red')
plt.title('plot 3')


plt.subplot(2,2,4)
plt.plot(y2,y1,color='black')
plt.title('plot 4')
```

```python
## Bar plot
categories = ['A','B','C','D','E']
values = [5,3,2,6,8]

## create a Bar plot
plt.bar(categories,values,color='gray')

# plt.plot(x,y)
plt.xlabel('categories')
plt.ylabel('values')
plt.title('Basic Line plot')
plt.show()
```

### Histogram
A histogram is a graphical representation used in statistics to display the frequency distribution of continuous numerical data.

```python
## sample test

data = [1,2,2,3,3,3,4,4,4,4,5,5]

##create a histogram

plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
## create a scatter plot
# Sample data

x = [1,2,3,4,5]
y = [2,3,4,5,6]

plt.scatter(x,y,color="blue",marker='x')
```

```python
## pie chart
labels = ['A','B','C','D']
sizes = [20,50,10,20]
colors = ['red','gold','lightskyblue','green']
explode = (0.1,0,0,0) ## move the first slice of the pie chart

## create a pie chart
plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct='%1.1f%%')
```

```python
## sales data visualization

import pandas as pd
sales_data_df = pd.read_csv('Chocolate_Sales.csv')
sales_data_df.head(5)
```

```python
sales_data_df.info()
```

```python
## plot total sales by product

total_sales_data_df = sales_data_df.groupby('Country')['Boxes_Shipped'].sum().head(5)
print(total_sales_data_df)
```

```python
total_sales_data_df.plot(kind='bar',color='teal')
```

```python
## plot total sales by product
sales_trend=sales_data_df.groupby('Order_Date')['Boxes_Shipped'].sum().reset_index().head(100)
plt.plot(sales_trend['Order_Date'],sales_trend['Boxes_Shipped'])
```
