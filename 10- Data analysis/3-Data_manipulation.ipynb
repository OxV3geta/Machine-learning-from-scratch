```python
## Handling Missing values
df.isnull().any(axis=1) ## True = Missing value
```

```python
df.isnull().sum()
```

```python
df_filled=df.fillna(0)
```

```python
### filling missing value with the mean of the 

df['Price_per_Box_fillNA'] = df['Price_per_Box'].fillna(df['Price_per_Box'].mean())
df.head()
```

```python
df.dtypes
```

```python
# Renames Columns

df = df.rename(columns={'Order_Date':'Order_day'})
df.head()
```

```python
## change data types
# df.dtypes
# df['Discount_Pct_new'] = df['Discount_Pct'].fillna(df['Discount_Pct'].mean()).astype(int)
# df.head(5)
df['Discount_Pct_new'] = df['Discount_Pct'].fillna(df['Discount_Pct'].mean()).astype(int)
df.head(5)
```

```python
# df['New_Discount_Pct'] = df['Discount_Pct'].apply(lambda x:x*2)
# df
df['New_Discount_Pct'] = df['Discount_Pct'].apply(lambda x:x*2)
df.head()
```

```python
df.head(5)
```

```python
 ## determine mean based on the something common like same_product mean,same things etccc
grouped_mean = df.groupby('Product')['Marketing_Spend'].mean()
print(grouped_mean)
```

```python
## grouped of product sum,mean etccc
grouped_sum = df.groupby(['Country','Product'])['Boxes_Shipped'].sum()
print(grouped_sum)
```

```python
df.groupby(['Country','Product'])['Boxes_Shipped'].mean()
```

```python
##aggregate multiple functions 

grouped_agg = df.groupby('Country')['Price_per_Box'].agg(['mean','sum','count','sum'])
grouped_agg
```

```python
grouped_agg = df.groupby('Country')['Price_per_Box'].agg(['sum','mean','count'])
grouped_agg
```

```python
### merging and joining Dataframes
## Create sample DataFrames

df1 = pd.DataFrame({'key' : ['A','B','C'],'Value1':[1,2,3]})
df2 = pd.DataFrame({'key' : ['A','B','D'],'Value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
## Merge Dataframe on the 'key columns'

pd.merge(df1,df2,on="key",how='inner') ## inner will give only matching key value
```

```python
pd.merge(df1,df2,on="key",how='outer') ## outer will give all key's value,if there have no value then give you 'NaN'
```

```python
pd.merge(df1,df2,on="key",how='left')
```

```python
pd.merge(df1,df2,on="key",how='right')
```
