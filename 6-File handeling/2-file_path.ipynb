### File path

```python
## create a new directory
import os
new_directory = "package"
os.mkdir(new_directory)
print(f"Directory{new_directory} is created.")
```

```python
##  Listing Files and Directories
## '.'(dot) basically means we looking at the current working location
item = os.listdir()
print(item)
```

```python
### joining path

dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
path = "example1.txt"

if os.path.exists(path):
    print(f"the path '{path}' exist.")
else:
    print(f"the path '{path}' doesn't exist.")
```

```python
## check if a path is a file or Directory

import os

path = "example.txt"

if os.path.isfile(path):
    print(f"The path '{path}' is file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is directory.")
else:
    print(f"The path '{path}' is neither a directory nor a file.")
```
