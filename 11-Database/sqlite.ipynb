### SQL and SQLite

SQL (Structured Query Language) is a standard, domain-specific programming language used to manage, query, and manipulate data stored in relational database management systems (RDBMS).

SQLite is a self-contained, serverless, zero-configuration, transactional SQL database engine** implemented as a C-language library.  Unlike traditional client-server databases, it operates embedded within the host application, storing the entire database—including tables, indices, and triggers—in a single cross-platform disk file

```python
import sqlite3
```

```python
## Connect to an SQlite database

connection = sqlite3.connect('example.db')
connection
```

```python
## cursor will iterate the table

cursor = connection.cursor()
```

```python
## create a table
cursor.execute(
'''
    Create Table If Not Exists employees(
        id Integer Primary key,
        name Text not null,
        age Integer,
        department text
)
''')

## commit the changes
connection.commit()
```

```python
cursor.execute(
    '''
    Select * from employees
    '''
)
```

```python
## Insert the data in sqlite table
cursor.execute(
    '''
    Insert Into employees(name,age,department)
                    values('Vegeta',47,'data scientist')
    
    '''
)

cursor.execute(
    '''
    Insert Into employees(name,age,department)
                    values('goku',42,'Wev dev')
    
    '''
)

cursor.execute(
    '''
    Insert Into employees(name,age,department)
                    values('Gohan',22,'scholer')
    
    '''
)

##
connection.commit()
```

```python
## Query the data from the table 
cursor.execute('select * from employees')
rows = cursor.fetchall()

## print the queried data
for row in rows:
    print(row)
```

```python
## update the data in the table
cursor.execute(''' 
               
update employees
set age=50
where name='goku'

''')

connection.commit()
```

```python
## Query the data from the table 
cursor.execute('select * from employees')
rows = cursor.fetchall()

## print the queried data
for row in rows:
    print(row)
```

```python
## delete the data from the table
cursor.execute('''
delete from employees
        where name = 'Gohan'
''')

connection.commit()
```

```python
## Query the data from the table 
cursor.execute('select * from employees')
rows = cursor.fetchall()

## print the queried data
for row in rows:
    print(row)
```

```python
## Working with sales data
## connect to an SQlite database

connection = sqlite3.connect('Sales_data.db')
cursor = connection.cursor()

## create table for sales data
cursor.execute('''
create table if not exists sales(
    id integer primary key,
    date text not null,
    product text not null,
    sales intger,
    region text
)
''')

## Insert data into the sales table

sales_data = [
    ('2020-01-01','product1',100,'North'),
    ('2020-01-02','product2',200,'south'),
    ('2020-01-03','product3',300,'west'),
    ('2020-01-04','product4',400,'east'),
    ('2020-01-05','product5',500,'North')
] 
cursor.executemany('''
Insert into sales(date,product,sales,region)
                    values(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
cursor.execute('select * from sales')
rows =cursor.fetchall()

## print the queired data
for row in rows:
    print(row)
```
