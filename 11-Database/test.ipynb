```python
import sqlite3
```

```python
connection = sqlite3.connect('test.db')
connection
```

```python
cursor = connection.cursor()
```

```python
cursor.execute('''
create table if not exists Saiyan(
    id integer primary key,
    name text not null,
    age integer,
    path text
)
''')

connection.commit()
```

```python
cursor.execute('''
insert into Saiyan(name,age,path)
            values('Gohan',24,'Beast')
            
''')

cursor.execute('''
insert into Saiyan(name,age,path)
            values('Goku',44,'Angle')             
''')

connection.commit()
```

```python
cursor.execute('select * from Saiyan')
rows = cursor.fetchall()

for row in rows:
    print(row)
```
