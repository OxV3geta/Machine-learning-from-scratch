```python
##1
import sqlite3
connection = sqlite3.connect('test.db')
connection
```

```python

def create_table():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    create table if not exists employees(
    id integer primary key,
    name text not null,
    age integer,
    department text 
    )
    ''')
    conn.commit()
    conn.close()
    print("Table 'employees' created successfully.")
    
    
create_table()
    
```

```python
##2
def insert_employee(id,name,age,department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    insert into employees(id,name,age,department)
                    values(?,?,?,?)
    ''',(id,name,age,department))
    conn.commit()
    conn.close()
    print('Employee inserted successfully.')

insert_employee(1,'Goku',43,'HR')
```

```python
insert_employee(2, 'Bob', 25, 'Engineering')
insert_employee(3, 'Charlie', 28, 'Sales')
insert_employee(4, 'David', 35, 'Marketing')
insert_employee(5, 'Eve', 22, 'HR')
```

```python
## 3
def fetch_all_employees():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('select *from employees')
    records = cursor.fetchall()
    conn.close()
    for record in records:
        print(record)

fetch_all_employees()
```

```python
def fetch_employee_from_dep(department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('select *from employees where department = ?',(department,))
    records = cursor.fetchall()
    conn.close()
    for record in records:
        print(record)
        
fetch_employee_from_dep('HR')
```

```python
def update_dep(Emp_id,new_dep):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    update employees
    set department = ?
    where id =?
    ''',(new_dep,Emp_id))
    conn.commit()
    conn.close()
    print('Employee department updated successfully')

update_dep(1,'CEO')
```

```python
update_dep(3,'Manager')
update_dep(4,'HR')
fetch_all_employees()
```

```python
def delete(employee_id):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    delete from employees
    where id =?          
    ''',(employee_id,))
    conn.commit()
    conn.close()
    print("Employee deleted successfully.")

delete(4)
```

```python
def fetch_employee_older_than(age):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('select *from employees where age > ?',(age,))
    records = cursor.fetchall()
    conn.close()
    for record in records:
        print(record)

fetch_employee_older_than(23)
    
```

```python
def fetch_employee_name_start_with(Latter):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('select *from employees where name like ?',(Latter+'%',))
    records = cursor.fetchall()
    for record in records:
        print(record)

fetch_employee_name_start_with('B')
```

```python
def insert_multiple_employees(employees):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    try:
        cursor.executemany('''
        insert into employees (id,name,age,department)
        values(?,?,?,?)
        ''',employees)
        conn.commit()
        conn.close()
        print("All employees inserted successfully.")
    except Exception as e:
        conn.rollback()
        print("Error occurred,transection rolled back.")
        print(e)
    finally:
        conn.close()

employees = [
    (6, 'Frank', 40, 'Finance'),
    (7, 'Grace', 29, 'Engineering'),
    (8, 'Hannah', 35, 'Marketing'),
    (9, 'Ivan', 38, 'Sales'),
    # (6, 'Jack', 45, 'HR')
]
insert_multiple_employees(employees)
```

```python
def update_multiple_age(updates):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    try:
        cursor.executemany('''
        update employees
        set age = ?
        where id = ?
        ''',updates)
        conn.commit()
        conn.close()
        print("All employee ages updated successfully.")
    except Exception as e:
        conn.rollback ###n Python, rollback is a database transaction management operation that reverts all changes made during a transaction to its previous state, effectively undoing uncommitted modifications.
        print("Error occurred, transaction rolled back.")
        print(e)
    finally:
        conn.close
        
updates = [
    (37, 1),
    (26, 2),
    (33, 3),
    (49, 989),  # Non-existing ID to cause an error
    (23, 5)
]
update_multiple_age(updates)
```
