### Sets
### a set is a built-in data structure used to store a collection of items that are unordered, unique, and unindexed.

```python
## create a set
my_set = {1,2,3,4,5,6}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_set))
```

```python
my_empty_set = set([1,2,3,4,5,6])
print(type(my_empty_set))
```

```python
my_empty_set = set([1,2,3,4,4,5,5,6])
print((my_empty_set))
```

```python
## Basic set operation
##Adding and removing
my_set = {1,2,3,4,4,5,5,6}
my_set.add(7)
print(my_set)
```

```python
## removing 
my_set = {1,2,3,4,4,5,5,6}
my_set.remove(3)
print(my_set)
```

```python
my_set = {1,2,3,4,5,6}
my_set.discard(11)
print(my_set)
```

```python
## pop method
my_set = {1,2,3,4,5,6}
removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
## clear all the elements
my_set = {1,2,3,4,5,6}
my_set.clear()
print(my_set)
```

```python
##set Membership test
my_set = {1,2,3,4,5}
print(2 in my_set)
print(7 in my_set)
```

```python
##Mathmatical operation
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

##Union
union_set = set1.union(set2)
print(union_set)

#Intersection

intersection_set= set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Difference
updated_set = set1.difference(set2)
print(updated_set)
print(set2.difference(set1))
```

```python
##symmetric Difference
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}
set1.symmetric_difference(set2) ##common elements get remove,remaining unique elements get combine
```

```python
## sets method
set1 = {1,2,3}
set2 = {3,4,5}
set3 = {1,2,3,4}

print(set1.issubset(set3))
print(set3.issubset(set1))

print(set3.issuperset(set1))
```

```python
### Counting Unique words in text

text = "I'm Vegeta The prince of all Saiyan"
words=text.split(" ")

##convert list of world tp set to get unique words

unique_words=set(words)
print(unique_words)
print(len(unique_words))
 
```
