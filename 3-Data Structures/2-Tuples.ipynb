## Packing and unpacking Tuple
### Packing groups multiple items into a single container (like a tuple or list), while unpacking reverses this, assigning items from a container to individual variables.

```python
## Packing Tuple

packed_tuple = 1,"hello world",3.14
print(packed_tuple)
```

```python
## Unpacked Tuple
a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
## Unpacking with *

numbers = (1,2,3,4,5,6)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
## Nested Tuple
## Nested List

lst = [[1,2,3,4],[5,6,7,8,9],[1,"Hello world",3,1416]]
lst[1][0:3]
```

```python
nested_tuple = ((1,2,3),("a","b","c"),(True,False))

print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuple
nested_tuple = ((1,2,3),("a","b","c"),(True,False))

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=' ')
    print()
        
```
