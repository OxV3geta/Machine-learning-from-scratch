### Dictionary
### A Python dictionary is a built-in, mutable data structure that stores data as key-value pairs, allowing for efficient data retrieval using unique, immutable keys. 

```python
dict = {}
my_set = {1,2,3,4,5,6}
print(type(dict))
print(type(my_set))
```

```python
empty_dict=dict()
empty_dict
```

```python
student = {"name":"Vegeta","age":47,"grade":"S","name":"goku"}
print(student)
```

```python
##accessing Dictionary Elements
student = {"name":"Vegeta","age":47,"grade":"S"}

print(student['grade'])
print(student['name'])

## Accessing using get() method
print(student.get('age'))
print(student.get('Last_name'))

print(student.get('Last_name',"Not Available"))
```

```python
## Modifying Dictionary Elements
## Dictionary are mutable
student = {"name":"Vegeta","age":47,"grade":"S"}

student["age"] = 31
print(student)

student["address"] = "Earth"
print(student)

del student['grade']
print(student)
```

```python
## Dictionary methods
student = {"name":"Vegeta","age":47,"grade":"S"}

keys = student.keys() ##get all the keys
print(keys)
values=student.values() ##get all the values
print(values) 

items=student.items()
print(items)
```

```python
##Shallow copy
student = {"name":"Vegeta","age":47,"grade":"S"}

student_copy = student
print(student)
print(student_copy)

student["name"] = "goku"
print(student)
print(student_copy)
```

```python
student = {"name":"Vegeta","age":47,"grade":"S"}
student_copy1 = student.copy() ##shallow copy
print(student)
print(student_copy1)
```

```python
student = {"name":"Vegeta","age":47,"grade":"S"}
for key in student.keys():
    print(key)

print()

for item in student.values():
    print(item)
```

```python
student = {"name":"Vegeta","age":47,"grade":"S"}
for keys,items in student.items():
    print(f"{keys} : {items}")
```

```python
 ##Nested Dictionary
 
students = {
    "student1":{"name":"Vegeta","age":47},
    "student2":{"name":"Goku","age":43}
}
print(students)
```

```python
print(students["student1"]["name"])
print(students["student2"]["name"])
```

```python
## Iterationg nested dictionaries element
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key} : {value}")
```

```python
### Dictionary comphrehension
squeres={x:x**2 for x in range(5)}
print(squeres)
```

```python
evens = {x:x**2 for x in range(11) if x%2==0}
print(evens)
```

```python
## use dictionary to count the frequency of elements in the list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1

print(frequency)
```

```python
## Merge 2 dictionaries into one

dict1={"a":1 , "b":2}
dict2={"c":3 , "d":4}
merged_dict = {**dict1,**dict2}
print(merged_dict)
```
